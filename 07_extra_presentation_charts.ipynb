```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Data
data = {
    "Test set": ["250k abstracts"] * 3 + ["400k mixed"] * 3 + ["400k titles only"] * 3,
    "Model": ["SVM", "BERT", "Gemini"] * 3,
    "F1 Score": [0.962, 0.993, 0.866, 0.941, 0.948, 0.859, 0.863, 0.843, 0.839]
}
df = pd.DataFrame(data)

# Style setup
sns.set(style="whitegrid")
plt.figure(figsize=(12, 8))  # Keep original size

# Define order and color palette
model_order = ["SVM", "BERT", "Gemini"]
palette = {
    "SVM": "#1f77b4",
    "BERT": "#4a90e2",
    "Gemini": "#a7c7e7"
}

# Plot
ax = sns.barplot(
    data=df,
    x="Test set",
    y="F1 Score",
    hue="Model",
    hue_order=model_order,
    palette=palette
)

# Annotate with large, bold font above bars
for container in ax.containers:
    ax.bar_label(container, fmt="%.3f", padding=8, fontsize=18, weight='bold')

# Text styling for slides
ax.set_title("F1 Score by Model and Test Set", fontsize=26, fontweight='bold', pad=15)
ax.set_ylabel("F1 Score", fontsize=20, labelpad=10)
ax.set_xlabel("", fontsize=18)
ax.set_ylim(0.72, 1.01)
ax.tick_params(labelsize=18)

# Legend — larger and clear
ax.legend(
    title="Model",
    title_fontsize=20,
    fontsize=18,
    loc="upper right"
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Data
best_combo = pd.DataFrame({
    "Model": ["SVM", "BERT", "Gemini", "Voting Ensemble"],
    "Micro-F1": [0.954, 0.968, 0.940, 0.966],
    "Macro-F1": [0.902, 0.924, 0.869, 0.922]
})
df_melted = best_combo.melt(id_vars="Model", var_name="Metric", value_name="F1")

# Consistent model colors
model_colors = {
    "Micro-F1": "#4a90e2",   # Blue
    "Macro-F1": "#1f77b4"    # Darker blue
}

sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=df_melted, x="Model", y="F1", hue="Metric",
    palette=model_colors, edgecolor="black"
)
ax.set_title("Micro vs Macro-F1 for Best Model Variants", fontsize=22, fontweight="bold", pad=15)
ax.set_ylim(0.75, 1.0)
ax.set_ylabel("F1 Score", fontsize=18)
ax.set_xlabel("", fontsize=16)
ax.tick_params(labelsize=14)
ax.legend(title="Metric", title_fontsize=16, fontsize=14, loc="upper left", bbox_to_anchor=(1, 1))
for container in ax.containers:
    ax.bar_label(container, fmt="%.3f", padding=6, fontsize=14, weight="bold")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Data
best_combo = pd.DataFrame({
    "Model": ["SVM", "BERT", "Gemini", "Voting Ensemble"],
    "Micro-F1": [0.954, 0.968, 0.940, 0.966],
    "Macro-F1": [0.902, 0.924, 0.869, 0.922]
})
df_melted = best_combo.melt(id_vars="Model", var_name="Metric", value_name="F1")

# Set up style and base plot
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))

# Consistent blue color palette for metrics
palette = {
    "Micro-F1": "#4a90e2",
    "Macro-F1": "#1f77b4"
}

ax = sns.barplot(
    data=df_melted, x="Model", y="F1", hue="Metric",
    palette=palette, edgecolor="black"
)

# Overlay gray boxes on all bars except Voting Ensemble
bar_width = 0.4  # width of each bar within the grouped bars
offsets = {"Micro-F1": -bar_width/2, "Macro-F1": bar_width/2}

for i, row in df_melted.iterrows():
    if row["Model"] != "Voting Ensemble":
        xpos = list(df_melted["Model"].unique()).index(row["Model"]) + offsets[row["Metric"]]
        ax.add_patch(plt.Rectangle(
            (xpos - bar_width/2, 0),
            width=bar_width,
            height=row["F1"] ,
            color="#cccccc",
            zorder=3,
            alpha=0.85
        ))

# Titles, labels, legend
ax.set_title("Micro vs Macro-F1 for Best Model Variants", fontsize=22, fontweight="bold", pad=15)
ax.set_ylim(0.75, 1.0)
ax.set_ylabel("F1 Score", fontsize=18)
ax.set_xlabel("")
ax.tick_params(labelsize=14)
ax.legend(title="Metric", title_fontsize=16, fontsize=14, loc="upper left", bbox_to_anchor=(1, 1))

# Value labels
for container in ax.containers:
    ax.bar_label(container, fmt="%.3f", padding=6, fontsize=14, weight="bold")

plt.tight_layout()
plt.show()
```
